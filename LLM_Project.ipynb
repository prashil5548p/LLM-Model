```python
model = "google/flan-t5-small"
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained(model)
model = AutoModelForSeq2SeqLM.from_pretrained(model)
```

```python
context = '''My favourite bike is the Kawasaki Ninja Z900,
           It is one of the most stylish and powerful bikes,
           It has a very attractive and sporty design,
           Its engine is powerful and gives amazing performance.,
           The bike has a 948cc four-cylinder engine.,
           It produces a thrilling and exciting sound.
           I love its aggressive front look.
           Its green and black colour combination looks beautiful.
           The bike offers excellent speed and acceleration.
           It has a comfortable riding position.
           Its braking system provides good control and safety.
           The LED headlights give it a modern appearance.
           It is perfect for long rides and highway journeys.
           Riding a Kawasaki Ninja Z900 is my dream.
           I hope to buy this beautiful bike in the future.'''
```

```python
model.eval()
```

```python
import builtins
input = builtins.input
question = input("Ask me question - ")
```

```python
prompt = f'''
Answer the questions only using the data in context.
If information is not available in context
say : Information is not provided

file_name : {context}
question : {question}
answer :
'''
```

```python
inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=512,
    truncation=True,
)
```

```python
# generate answer
with torch.no_grad():
  output = model.generate(
      **inputs,
      max_new_tokens = 100,
      num_beams = 4
  )
```

```python
# decode answer
answer =tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
print('Question - ', question)
print('Answer - ', answer)
```
